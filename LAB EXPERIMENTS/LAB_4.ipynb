```python
def plural(word):
    if word.endswith(('s', 'x', 'ch', 'sh')):
        return word + "es"
    elif word.endswith('y') and word[-2] not in 'aeiou':
        return word[:-1] + "ies"
    else:
        return word + "s"

words = ["cat", "box", "baby", "bus"]

for w in words:
    print(w, "->", plural(w))
```
