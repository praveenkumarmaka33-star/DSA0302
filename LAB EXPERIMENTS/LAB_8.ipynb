```python
import nltk
from nltk import ConditionalFreqDist

nltk.download('brown')
nltk.download('universal_tagset')

from nltk.corpus import brown

data = brown.tagged_words(tagset='universal')

cfd = ConditionalFreqDist(
    (word.lower(), tag) for word, tag in data
)

def pos_tag(word):
    if word.lower() not in cfd:
        return "UNKNOWN"

    return cfd[word.lower()].max()

words = ["the", "run", "beautiful", "dog"]

for word in words:
    print(word, ":", pos_tag(word))
```
