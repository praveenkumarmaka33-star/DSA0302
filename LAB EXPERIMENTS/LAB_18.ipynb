```python
import re

def fopc_parser(expression):
    pattern = r'(forall|exists)\s+(\w+)\s*:\s*(\w+)\((\w+)\)'

    match = re.fullmatch(pattern, expression.strip())

    if match:
        print("Quantifier:", match.group(1))
        print("Variable:", match.group(2))
        print("Predicate:", match.group(3))
        print("Argument:", match.group(4))
    else:
        print("Invalid FOPC Expression")

fopc_parser("forall x : Human(x)")
```
