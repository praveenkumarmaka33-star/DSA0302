```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> SING | PLUR
SING -> NS VS
PLUR -> NP VP
NS -> 'he' | 'she'
VS -> 'runs' | 'eats'
NP -> 'they' | 'we'
VP -> 'run' | 'eat'
""")

parser = nltk.ChartParser(grammar)

sentences = [
    "he runs",
    "they run",
    "he run",
    "they runs"
]

for s in sentences:
    try:
        trees = list(parser.parse(s.split()))
        print(s, "->", "Correct" if trees else "Incorrect")
    except ValueError:
        print(s, "-> Incorrect")
```
