```python
import random

text = "I love natural language processing and I love Python"
words = text.split()

bigrams = list(zip(words, words[1:]))

print("Bigrams:", bigrams)

result = [words[0]]

for i in range(5):
    next_words = [b for a, b in bigrams if a == result[-1]]

    if not next_words:
        break

    result.append(random.choice(next_words))

print("Generated Text:", " ".join(result))
```
