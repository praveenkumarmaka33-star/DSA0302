```python
import nltk

from nltk.tag import RegexpTagger, UnigramTagger, BrillTaggerTrainer
from nltk.corpus import treebank

nltk.download('treebank')

train = treebank.tagged_sents()[:2000]

initial = RegexpTagger([
    (r'.*ing$', 'VBG'),
    (r'.*ed$', 'VBD'),
    (r'.*', 'NN')
])

templates = [
    nltk.tag.brill.Template(
        nltk.tag.brill.Pos([-1])
    )
]

trainer = BrillTaggerTrainer(initial, templates)

tagger = trainer.train(train, max_rules=10)

print(tagger.tag("The boy is playing football".split()))
```
