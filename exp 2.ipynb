```python
n, k = map(int, input().split())

alphabet = input().split()

start = int(input())

data = list(map(int, input().split()))
m = data[0]
accepting = set(data[1:])

t = int(input())

trans = {}

for _ in range(t):
    p, c, r = input().split()
    p = int(p)
    r = int(r)

    trans.setdefault((p, c), set()).add(r)

q = int(input())

for _ in range(q):
    s = input()

    if s == "#":
        s = ""

    current = {start}

    for c in s:
        nxt = set()

        for state in current:
            nxt.update(trans.get((state, c), set()))

        current = nxt

        if not current:
            break

    if current & accepting:
        print("ACCEPT")
    else:
        print("REJECT")
```
