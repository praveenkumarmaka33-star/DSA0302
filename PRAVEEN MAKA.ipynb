```python
n, k = map(int, input().split())

alphabet = input().split()

start = int(input())

data = list(map(int, input().split()))
m = data[0]
accepting = set(data[1:])

t = int(input())

trans = {}

for _ in range(t):
    p, c, r = input().split()
    p = int(p)
    r = int(r)

    trans.setdefault((p, c), set()).add(r)

q = int(input())

for _ in range(q):
    s = input()

    if s == "#":
        s = ""

    current = {start}

    for c in s:
        nxt = set()

        for state in current:
            nxt.update(trans.get((state, c), set()))

        current = nxt

        if not current:
            break

    if current & accepting:
        print("ACCEPT")
    else:
        print("REJECT")
```

```python
VOWELS = set("aeiou")


def doublable(word):
    if len(word) < 3:
        return False

    a, b, c = word[-3:]

    # consonant, vowel, consonant
    if a in VOWELS or b not in VOWELS or c in VOWELS:
        return False

    if c in "wxy":
        return False

    # exactly one vowel group
    groups = 0
    in_vowel_group = False

    for ch in word:
        if ch in VOWELS:
            if not in_vowel_group:
                groups += 1
                in_vowel_group = True
        else:
            in_vowel_group = False

    return groups == 1


def generate(stem, feature):

    # Rule A
    if feature in ("N+PL", "V+3SG"):
        if (stem.endswith("s") or stem.endswith("x") or
            stem.endswith("z") or stem.endswith("ch") or
            stem.endswith("sh")):
            return stem + "es"

        if (len(stem) >= 2 and
            stem[-1] == "o" and stem[-2] not in VOWELS):
            return stem + "es"

        if (len(stem) >= 2 and
            stem[-1] == "y" and stem[-2] not in VOWELS):
            return stem[:-1] + "ies"

        return stem + "s"

    # Rule B
    if feature == "V+PROG":
        if (len(stem) >= 3 and
            stem.endswith("e") and
            stem[-2] not in VOWELS):
            return stem[:-1] + "ing"

        if doublable(stem):
            return stem + stem[-1] + "ing"

        return stem + "ing"

    # Rule C
    if feature in ("V+PAST", "ADJ+COMP", "ADJ+SUP"):

        suffix = {
            "V+PAST": "ed",
            "ADJ+COMP": "er",
            "ADJ+SUP": "est"
        }[feature]

        if stem.endswith("e"):
            if feature == "V+PAST":
                return stem + "d"
            elif feature == "ADJ+COMP":
                return stem + "r"
            else:
                return stem + "st"

        if (len(stem) >= 2 and
            stem[-1] == "y" and
            stem[-2] not in VOWELS):
            return stem[:-1] + "i" + suffix

        if doublable(stem):
            return stem + stem[-1] + suffix

        return stem + suffix


L = int(input())

lexicon = {}

for _ in range(L):
    stem, feature, form = input().split()
    lexicon[(stem, feature)] = form

Q = int(input())

for _ in range(Q):
    stem, feature = input().split()

    if (stem, feature) in lexicon:
        print(lexicon[(stem, feature)])
    else:
        print(generate(stem, feature))
```
