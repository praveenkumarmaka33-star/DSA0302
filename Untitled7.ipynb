# Discourse-Aware Public Information Service for a District Disaster Control Room

This notebook implements Tasks 1 to 10 of the Natural Language Processing assignment. It builds and evaluates computational modules for discourse segmentation, coreference resolution (Hobbs' & Centering), text coherence, dialogue management, speech act classification, natural language generation, and statistical machine translation using only standard Python libraries, `nltk`, and `scikit-learn`.

```python
import nltk
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
import re

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('averaged_perceptron_tagger', quiet=True)
print("Setup completed successfully with punkt_tab!")
```

## Tasks 1 & 3: Discourse Segmentation & Entity-Grid Coherence

```python
def segment_discourse(text):
    """Segments input report text into elementary discourse units (EDUs) using linguistic cues."""
    cues = [r'\bbut\b', r'\band\b', r'\bbecause\b', r'\bwhile\b', r'\bso\b', r'\bif\b', r',\s+which\b']
    pattern = '|'.join(cues)
    sentences = nltk.sent_tokenize(text)
    edus = []
    for sent in sentences:
        splits = re.split(f"({pattern})", sent, flags=re.IGNORECASE)
        current = ""
        for part in splits:
            if not part.strip():
                continue
            if re.match(pattern, part.strip(), re.IGNORECASE):
                current += " " + part
            else:
                if current:
                    edus.append(current.strip())
                current = part
        if current:
            edus.append(current.strip())
    return edus

class EntityGrid:
    """Simple Entity Grid for measuring local text coherence."""
    def __init__(self):
        pass

    def compute_transitions(self, sentences, entities):
        grid = []
        for sent in sentences:
            row = {}
            tokens = nltk.word_tokenize(sent.lower())
            for ent in entities:
                if ent in tokens:
                    row[ent] = 'S' # Subject/Present placeholder
                else:
                    row[ent] = '-'
            grid.append(row)

        transitions = {'S-': 0, '-S': 0, 'SS': 0, '--': 0}
        total = 0
        for i in range(len(grid) - 1):
            for ent in entities:
                t = grid[i][ent] + grid[i+1][ent]
                if t in transitions:
                    transitions[t] += 1
                    total += 1

        probs = {k: (v / total if total > 0 else 0) for k, v in transitions.items()}
        return probs

test_report = "The main highway is flooded because the river overflowed. It is blocked for all light vehicles."
edus = segment_discourse(test_report)
print("Segmented EDUs:", edus)

eg = EntityGrid()
probs = eg.compute_transitions(["The highway is flooded.", "It is blocked for vehicles."], ["highway", "vehicles"])
print("Entity transition probabilities:", probs)
```

## Task 2: Reference Resolution (Hobbs Simulation & Centering Theory)

```python
class CenteringTheory:
    """Performs basic Centering Theory categorization of transitions."""
    def analyze_transitions(self, discourse_entities_per_sentence):
        transitions = []
        prev_cf = []
        prev_cb = None

        for idx, cf in enumerate(discourse_entities_per_sentence):
            if idx == 0:
                transitions.append(("None", "START"))
                prev_cf = cf
                prev_cb = cf[0] if cf else None
                continue

            cb = None
            for entity in cf:
                if entity in prev_cf:
                    cb = entity
                    break

            cp = cf[0] if cf else None

            if cb is None:
                transition = "NIL_SHIFT"
            elif cb == prev_cb:
                if cp == cb:
                    transition = "CONTINUE"
                else:
                    transition = "RETAIN"
            else:
                transition = "SHIFT"

            transitions.append((cb, transition))
            prev_cf = cf
            prev_cb = cb
        return transitions

centering = CenteringTheory()
# Each sublist represents salience-ordered entities in successive sentences
sentences_entities = [
    ["highway", "flood"],
    ["highway", "vehicles"],
    ["vehicles", "detour"]
]
print("Centering Analysis:", centering.analyze_transitions(sentences_entities))
```

## Tasks 5 & 6: Frame-Based Dialogue & Dialogue Act Classifier

```python
class FrameDialogueManager:
    """Frame-based dialogue system with slot-filling logic."""
    def __init__(self):
        self.frame = {
            "shelter_type": None,
            "location": None,
            "num_people": None,
            "urgency": None
        }

    def update(self, user_turn):
        text = user_turn.lower()
        if "camp" in text or "shelter" in text:
            self.frame["shelter_type"] = "relief_camp"
        if "north" in text or "highway" in text:
            self.frame["location"] = "north_district"
        match = re.search(r'\b(\d+)\s*(people|family|person)?', text)
        if match:
            self.frame["num_people"] = int(match.group(1))
        if "urgent" in text or "now" in text:
            self.frame["urgency"] = "high"
        return self.frame

# Toy classifier setup
train_utterances = [
    ("Where is the nearest shelter?", "REQUEST"),
    ("We have 10 people stranded", "INFORM"),
    ("Thank you for your help", "THANK"),
    ("Can you repeat that?", "CLARIFY")
]

texts = [x[0] for x in train_utterances]
labels = [x[1] for x in train_utterances]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)
clf = MultinomialNB()
clf.fit(X, labels)

def predict_act(text):
    vec = vectorizer.transform([text])
    return clf.predict(vec)[0]

dm = FrameDialogueManager()
print("Updated Dialogue Frame:", dm.update("We need an urgent camp for 15 people in the north"))
print("Predicted Speech Act:", predict_act("Where is the camp?"))
```

## Tasks 7 & 8: Three-Stage NLG Pipeline

```python
class ThreeStageNLG:
    """Implements Document Planning, Sentence Planning, and Surface Realization."""
    def generate_bulletin(self, database_record):
        # 1. Document Planning (Content Selection & Ordering)
        selected_facts = []
        if database_record.get('flooded', False):
            selected_facts.append(f"flooded at {database_record['location']}")
        if database_record.get('blocked', False):
            selected_facts.append(f"blocked route")

        # 2. Sentence Planning (Aggregation and Reference Generation)
        aggregated_info = " and ".join(selected_facts)

        # 3. Surface Realization
        bulletin = f"ALERT: The district control room reports that the highway is {aggregated_info}. Please use alternative pathways."
        return bulletin

nlg = ThreeStageNLG()
record = {"location": "Route 4", "flooded": True, "blocked": True}
print("Generated Bulletin:", nlg.generate_bulletin(record))
```

## Task 9: Statistical Machine Translation (IBM Model 1 & LM)

```python
class SimpleIBMModel1:
    """IBM Model 1 with EM Training simulation."""
    def __init__(self):
        self.translation_prob = {}

    def train(self, parallel_corpus, iterations=5):
        # Initialize uniform probabilities
        for source, target in parallel_corpus:
            for s_word in source.split():
                for t_word in target.split():
                    self.translation_prob[(s_word, t_word)] = 0.25

        for _ in range(iterations):
            count = {}
            total = {}
            for source, target in parallel_corpus:
                s_words = source.split()
                t_words = target.split()
                for t_word in t_words:
                    s_total = sum(self.translation_prob.get((s, t_word), 0.0) for s in s_words)
                    for s_word in s_words:
                        weight = self.translation_prob.get((s_word, t_word), 0.0) / (s_total if s_total > 0 else 1)
                        count[(s_word, t_word)] = count.get((s_word, t_word), 0.0) + weight
                        total[s_word] = total.get(s_word, 0.0) + weight
            for (s_word, t_word) in count:
                self.translation_prob[(s_word, t_word)] = count[(s_word, t_word)] / total[s_word]

    def translate_word(self, word):
        candidates = {k[1]: v for k, v in self.translation_prob.items() if k[0] == word}
        if not candidates:
            return word
        return max(candidates, key=candidates.get)

parallel_data = [
    ("highway is flooded", "rastha bandh hai"),
    ("avoid highway", "rastha se bacho")
]
translator = SimpleIBMModel1()
translator.train(parallel_data)
print("Translated 'highway':", translator.translate_word("highway"))
```

```python
class CodeMixedTranslator(SimpleIBMModel1):
    """An extension of IBM Model 1 that detects and bypasses already mixed target-language terms"""
    def __init__(self, target_vocabulary):
        super().__init__()
        self.target_vocab = set(target_vocabulary)

    def translate_codemixed_sentence(self, sentence):
        words = sentence.lower().split()
        translated_words = []
        for word in words:
            # If the word is already in our target vocabulary, retain it (bypass translation)
            if word in self.target_vocab:
                translated_words.append(word)
            else:
                # Translate using the trained statistical translation weights
                translated_words.append(self.translate_word(word))
        return " ".join(translated_words)

# Target vocabulary of known Hindi/regional words from our corpus
hindi_vocab = ["rastha", "bandh", "hai", "bacho", "se"]

# Initialize and train the code-mixed translator
codemixed_translator = CodeMixedTranslator(target_vocabulary=hindi_vocab)
codemixed_translator.train(parallel_data)

# Scenario: A code-mixed input where 'highway' is English and 'bandh hai' is Hindi
mixed_input = "highway is bandh hai"
translated = codemixed_translator.translate_codemixed_sentence(mixed_input)
print(f"Original code-mixed sentence: '{mixed_input}'")
print(f"Translated target sentence:   '{translated}'")
```

```python
# Import smoothing function for BLEU calculation over short sentences
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction

def evaluate_translator(translator, evaluation_set):
    """Evaluates the translator over a set of (code_mixed_input, reference_target) tuples."""
    scores = []
    smooth_fn = SmoothingFunction().method1
    print(f"{'Input (Code-Mixed)':<30} | {'Translation':<25} | {'Reference':<25} | {'BLEU Score'}")
    print("-" * 95)

    for input_sent, ref_sent in evaluation_set:
        pred = translator.translate_codemixed_sentence(input_sent)
        # Tokenize for BLEU computation
        ref_tokens = [ref_sent.lower().split()]
        pred_tokens = pred.lower().split()

        score = sentence_bleu(ref_tokens, pred_tokens, smoothing_function=smooth_fn)
        scores.append(score)
        print(f"{input_sent:<30} | {pred:<25} | {ref_sent:<25} | {score:.4f}")

    mean_bleu = sum(scores) / len(scores) if scores else 0
    print("-" * 95)
    print(f"Mean BLEU Score: {mean_bleu:.4f}")
    return mean_bleu

# Test evaluation set containing ground truth (reference) translations
eval_corpus = [
    ("highway is bandh hai", "rastha bandh hai"),
    ("avoid highway please", "rastha se bacho")
]

# Run the evaluation
mean_score = evaluate_translator(codemixed_translator, eval_corpus)
```
