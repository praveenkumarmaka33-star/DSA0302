```python
import re

text = "Python is easy. Python is powerful."

# Match pattern
print(re.match("Python", text))

# Search pattern
print(re.search("easy", text))

# Find all occurrences
print(re.findall("Python", text))

# Replace text
print(re.sub("Python", "Java", text))
```
